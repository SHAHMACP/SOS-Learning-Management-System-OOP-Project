# **Synthetic Dataset for Risk Assessment and Loan Approval Modeling**
This synthetic dataset comprises 20,000 records of personal and financial data, designed to facilitate the development of predictive models for risk assessment. 

## Problem Statement

**A bank aims to develop a data-driven credit risk assessment system to improve the loan approval process and minimize the risk of loan defaults.** The bank has provided a dataset containing applicants’ demographic, employment, income, credit history, existing debt, and other financial information.

The objective of this project is to analyze these factors and develop **machine learning models for two key banking tasks**:

1. **Risk Score Prediction (Regression):** Predict a continuous risk score for each loan applicant to assess their likelihood of financial instability or loan default.
2. **Loan Approval Prediction (Binary Classification):** Predict whether a loan application is likely to be **approved or denied** based on the applicant's financial and credit profile.

The project will also involve identifying the key factors influencing credit risk and loan approval, thereby providing the bank with **data-driven insights to support consistent, efficient, and risk-aware lending decisions.**


> The bank seeks to develop a machine learning-based credit risk assessment system that can evaluate loan applicants based on their demographic, employment, income, credit, and financial characteristics. The project aims to predict applicants' risk scores and loan approval outcomes, while identifying the key factors influencing these decisions. The resulting models can support the bank in making faster, more consistent, and data-driven lending decisions while managing potential default risk.
